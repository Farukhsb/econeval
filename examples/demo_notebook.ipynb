# EconEval Quick Demo

This notebook walks through the smallest useful EconEval flow:

1. load the example model and config
2. run the CLI against the `basic_model` example
3. inspect the generated JSON summary

Open this notebook from the repository root so the relative paths resolve correctly.

```python
import json
from pathlib import Path

from econeval.cli import run_cli
```

```python
root = Path.cwd()
config_path = root / "examples" / "basic_model" / "econeval.yml"
model_path = root / "examples" / "basic_model" / "model.py"
report_path = root / "examples" / "demo-report.json"

exit_code = run_cli(
    [
        "--config",
        str(config_path),
        "--model",
        str(model_path),
        "--class",
        "DemoModel",
        "--report",
        str(report_path),
    ]
)

payload = json.loads(report_path.read_text(encoding="utf-8"))
print("exit_code:", exit_code)
print(payload["summary"])
```

If you want to adapt this demo, copy the `basic_model` folder, change the YAML rules, and point the CLI at your own model class.
